```python
import csv

with open(r"../data/Restaurant_Reviews.tsv", "r", newline="", encoding="utf-8") as tsv_file:
    reader = csv.reader(tsv_file, delimiter="\t")
    with open(r"../data/Restaurant_Reviews.csv", "w", newline="", encoding="utf-8") as csv_file:
        writer = csv.writer(csv_file)
        writer.writerows(reader)

print("Conversion done!")
```
